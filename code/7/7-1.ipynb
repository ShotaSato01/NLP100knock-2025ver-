```python
with open('./SST-2/train.tsv', 'r') as f:
    lines = f.readlines()
lines = lines[1:]  
ls = []

for l in lines:
    dic = {}
    feature = {}
    text = l.split('\t')[0]
    label = l.split('\t')[1].rstrip('\n')
    dic["text"] = text
    dic["label"] = label

    #特徴ベクトルの計算
    for word in text.split():
        if word in feature:
            feature[word] += 1
        else:
            feature[word] = 1
    
    dic["feature"] = feature

    ls.append(dic)


print('===train.tsv===')
print(ls[0])
with open('./train_features.txt','a') as f:
    for i in ls:
       print(i,file=f) 
```
