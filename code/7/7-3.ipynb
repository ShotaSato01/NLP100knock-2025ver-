```python
import ast
import numpy as np
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

# 1. データの読み込み
texts = []
labels = []

with open("train_features.txt", "r", encoding="utf-8") as f:
    for line in f:
        item = ast.literal_eval(line.strip())
        texts.append(item['text'])
        labels.append(int(item['label']))

# 2. Bag of Words ベクトル化
vectorizer = CountVectorizer()
X = vectorizer.fit_transform(texts)
y = np.array(labels)

print(f"X shape: {X.shape}")
print(f"y shape: {y.shape}")

# 3. ロジスティック回帰で分類
clf = LogisticRegression()
clf.fit(X, y)

```

```python
dev_texts = []
dev_labels = []

with open("dev_features.txt", "r", encoding="utf-8") as f:
    for line in f:
        item = ast.literal_eval(line.strip())
        dev_texts.append(item["text"])
        dev_labels.append(int(item["label"]))

X_dev = vectorizer.transform(dev_texts)  # ← fitしない！transformだけ！
y_dev = np.array(dev_labels)

# ---------- 5. モデルによる予測と精度評価 ----------
y_pred = clf.predict(X_dev)
accuracy = accuracy_score(y_dev, y_pred)

# ---------- 6. 結果の出力 ----------
for i in range(len(y_dev)):
    print(f"[{i}] y_true: {y_dev[i]}, y_pred: {y_pred[i]}")

print(f"Dev Accuracy: {accuracy:.4f}")
```
