```python
with open('./SST-2/train.tsv', 'r') as f:
    lines = f.readlines()
lines = lines[1:]  
dic = {}

for l in lines:
    label = l.split('\t')[1].rstrip('\n')
    if label not in dic:
        dic[label] = 1
    else:
        dic[label] += 1
print('===train.tsv===')
print(dic)
print('positive ; ',dic['1'])
print('negative ; ',dic['0'])
```
