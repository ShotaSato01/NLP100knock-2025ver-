```python
with open('../../code/2/popular-names.txt') as f:
    content = f.readlines()
length = len(content)
dic = {}

for line in content:
    name = line.split("\t")[0]

    if name not in dic:
        dic[name] = 1
    else:
        dic[name] +=1 

for name in sorted(dic):
    print(dic[name],name)
```
