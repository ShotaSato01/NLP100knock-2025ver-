```python
N = 10
with open('../../code/2/popular-names.txt') as f:
    content = f.readlines()

length = len(content)
num_lines = length//N

start_idx = 0
for i in range(N):
    with open(f"../../code/2/2-5/python/output_{i}.txt","a") as f_:
        for j in range(num_lines):
            f_.write(content[start_idx+j])
    start_idx += j+1
```
