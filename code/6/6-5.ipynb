```python
#自然言語処理用のgensimパッケージ
from gensim.models import KeyedVectors
file = '../../code/6/GoogleNews-vectors-negative300.bin'
#fileをword2vec形式で読み込み
model = KeyedVectors.load_word2vec_format(file,binary = True)
```

```python
#カテゴリ名が'gram'から始まるものが文法的アナロジー、それ以外のものが意味的アナロジー
output = '../../code/6/6-4output.txt'

sem_cnt = 0
sem_true = 0

with open(output, 'r', encoding='utf-8') as f:
    for row in f:
        cols = row.split('\t')                      
        target = cols[1].split(' ')[3]                            
        pred = cols[3]                                              
        sem_cnt += 1                                            
        if target == pred:                                      
            sem_true += 1                                       

print('意味的アナロジーの正解率: {}'.format(sem_true / sem_cnt))
```
