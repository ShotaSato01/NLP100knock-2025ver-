```python
#自然言語処理用のgensimパッケージ
from gensim.models import KeyedVectors
file = '../../code/6/GoogleNews-vectors-negative300.bin'
#fileをword2vec形式で読み込み
model = KeyedVectors.load_word2vec_format(file,binary = True)
```

```python
from tqdm import tqdm
file2 = '../../code/6/capital-common-countries.txt'
output = '../../code/6/6-4output.txt'

with open(file2, 'r', encoding='utf-8') as f1:
    content = f1.readlines()

category = ''
with open(output, 'w', encoding='utf-8') as f2 :
    for row in tqdm(content, total=len(content)):                                                                           #tqdmで進捗を表示                                                                                               #各カテゴリー内での処理
            cols = row.rstrip().split()                                                                         #各行を空白区切りで行列に保存
            word, similarity = model.most_similar(positive=[cols[1], cols[2]], negative=[cols[0]], topn=1)[0]   #2,3行目を加算し、1行目を減算し、最も類似度の大きい単語とその類似度を保存                                                              #類似度の大きい単語とその類似度を表示
            f2.write('{}\t{}\t{}\t{}\t{}\n'.format(category, row.rstrip(),"|", word, similarity))   
```
