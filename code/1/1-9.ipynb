```python
import random
S = "I couldn’t believe that I could actually understand what I was reading : the phenomenal power of the human mind"
print(S)

S = list(S.split(" "))
for word in S:
    if len(word) <= 4:
        print(word,end= " ")
    else:
        ls = random.sample(range(2,len(word)),len(word)-2)
        word_ = ""
        word_ += word[0]
        for i in ls:
            word_ += word[i-1]
        word_ += word[len(word)-1]
        print(word_,end = " ")
```
