```python
S1 = "paraparaparadise"
S2 = "paragraph"

def char_n_gram(n,S):
    output = []
    length = len(S)
    for i in range(length-n+1):
        tmp = []
        for j in range(i,i+n):
            tmp.append((S[j]))
        output.append("".join(tmp))
    
    return set(output)

X = char_n_gram(2,S1)
Y = char_n_gram(2,S2)

print("X : ",X)
print("Y : ",Y)

print("sum : ",X|Y)
print("mul : ",X&Y)
print("sub : ",X-Y)

target = "se"
if target in X:
    print(f"'{target}' is in X")
else:
    print(f"'{target}' is not in X" )

if target in Y:
    print(f"'{target}' is in Y")
else:
    print(f"'{target}' is not in Y" )
```
