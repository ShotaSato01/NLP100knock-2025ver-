```python
S = input()

def cipher(S):
    output = ""
    for i in S:
        num = ord(i)
        if 97 <= num <= 122:
            tmp = 219 - num
            output += chr(tmp)
        else:
            output += i
    return output

def decryption(S):
    output = ""
    for i in S:
        num = ord(i)
        if 97 <= num <= 122:
            tmp = 219 - num
            output += chr(tmp)
        else:
            output += i
    return output

if __name__ == '__main__':
    cip = cipher(S)
    print(cip)
    dec = decryption(cip)
    print(dec)

#ord(S)：1 文字の Unicode 文字を表す文字列に対し、その文字の Unicode コードポイントを表す整数を返します。
```
