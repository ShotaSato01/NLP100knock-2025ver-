```python
#S = "I am an NLPer"
n = int(input("please input n >> "))
S = input("please input S >> ")

def char_n_gram(n,S):
    output = []
    length = len(S)
    for i in range(length-n+1):
        tmp = []
        for j in range(i,i+n):
            tmp.append((S[j]))
        output.append("".join(tmp))
    
    return output

def word_n_gram(n,S):
    S = S.split(" ")
    output = []
    length = len(S)
    for i in range(length-n+1):
        tmp = []
        for j in range(i,i+n):
            tmp.append((S[j]))
        output.append(tmp)
    
    return output

print(char_n_gram(n,S))
print(word_n_gram(n,S))
```
