```python
import MeCab

text = """
メロスは激怒した。
必ず、かの邪智暴虐の王を除かなければならぬと決意した。
メロスには政治がわからぬ。
メロスは、村の牧人である。
笛を吹き、羊と遊んで暮して来た。
けれども邪悪に対しては、人一倍に敏感であった。
"""

def tokenize_with_pos(text):
    mecab = MeCab.Tagger()
    nodes = mecab.parseToNode(text)
    tokens = []
    while nodes:
        if nodes.feature.split(',')[0] == "名詞" and nodes.next.surface == "の" and nodes.next.next.feature.split(',')[0] == "名詞":
            tokens.append(nodes.surface + "の" +nodes.next.next.surface)
        nodes = nodes.next
    return tokens

# 使用例
tokens_with_pos = tokenize_with_pos(text)
for words in tokens_with_pos:
    print(words)
```
