```python
import CaboCha
from graphviz import Digraph

text = """
メロスは激怒した。
"""

# CaboCha で係り受け解析
c = CaboCha.Parser()
tree = c.parse(text)

# グラフオブジェクト作成（左から右に描画）
dot = Digraph(format='png')
dot.attr('node', shape='box', fontname='IPAexGothic')  # 日本語フォント（Linux環境などで調整して）

# 文節ごとの文字列リスト
phrases = []

# 各文節を取得し、ノードとして追加
for i in range(tree.chunk_size()):
    chunk = tree.chunk(i)
    phrase = ''
    for j in range(chunk.token_pos, chunk.token_pos + chunk.token_size):
        phrase += tree.token(j).surface
    phrases.append(phrase)
    dot.node(str(i), phrase)

# 係り受けの矢印を追加
for i in range(tree.chunk_size()):
    chunk = tree.chunk(i)
    if chunk.link != -1:
        dot.edge(str(i), str(chunk.link))

# グラフを描画（ファイル名: "result"）
dot.render('4-5result', view=True)
```
