```python
import CaboCha

text = """
メロスは激怒した。
必ず、かの邪智暴虐の王を除かなければならぬと決意した。
メロスには政治がわからぬ。
メロスは、村の牧人である。
笛を吹き、羊と遊んで暮して来た。
けれども邪悪に対しては、人一倍に敏感であった。
"""

c = CaboCha.Parser()
tree = c.parse(text)

# 各チャンク（文節）の係り先と文節内容を抽出
for i in range(tree.chunk_size()):
    flg = 0
    chunk = tree.chunk(i)
    dst = chunk.link  # 係り先インデックス

    # 係り元の文節
    src_phrase = ''
    for j in range(chunk.token_pos, chunk.token_pos + chunk.token_size):
        if tree.token(j).surface == "メロス":
            flg = 1
        src_phrase += tree.token(j).surface

    # 係り先の文節（-1は係り先なし）
    if dst != -1 and flg == 1:
        dst_chunk = tree.chunk(dst)
        dst_phrase = ''
        for j in range(dst_chunk.token_pos, dst_chunk.token_pos + dst_chunk.token_size):
            dst_phrase += tree.token(j).surface

        print(f"{src_phrase}\t{dst_phrase}")
```
