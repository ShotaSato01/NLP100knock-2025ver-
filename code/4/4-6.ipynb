```python
import json
import re

lst=[]
with open ('../../code/3/jawiki-country.json','r',encoding='utf-8_sig') as f:
    for line in f.readlines():
        dic=json.loads(line)
        lst.append(dic)

texts = []
for content in lst:
    texts.append(content["text"])

output_texts = []
for t in texts:
    line = t.replace("\n","")
    output_texts.append(re.sub(r'\'*|\[*|\]*|{*|}*|=*|\||<!--(.*?)-->|~~~~|:|;|#',"",line))

with open("../../code/4/corpus.txt",'a') as f_:
    for text in output_texts:
        print(text,file=f_)
```

```python
import MeCab
from collections import Counter

# ファイル読み込み
with open("../../code/4/kokoro.txt", "r") as f:
    content = f.read()

# MeCabの解析器
tagger = MeCab.Tagger()

# （単語, 品詞）ペアを格納
word_pos_list = []

# 形態素解析
content_ = content.split('\n')
for country in content_:
    node = tagger.parseToNode(country)
    while node:
        surface = node.surface
        features = node.feature.split(',')
        pos = features[0]  # 品詞

        # 記号や助詞・助動詞などを除く
        if pos not in ['補助記号', '記号', '助詞', '助動詞', '接続詞']:
            if surface:
                word_pos_list.append((surface, pos))

        node = node.next

# 頻度カウント（単語 + 品詞 のタプルで集計）
counter = Counter(word_pos_list)

# 上位20件を出力
for (word, pos), freq in counter.most_common(20):
    print(f"{word}\t{pos}\t{freq}")
```
