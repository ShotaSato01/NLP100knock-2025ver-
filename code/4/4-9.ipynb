```python
import MeCab
import matplotlib.pyplot as plt
from collections import Counter
import numpy as np

# MeCab分かち書き用
tagger = MeCab.Tagger("-Owakati")

# コーパス読み込み
with open("tokenized_kokoro.txt", "r", encoding="utf-8") as f:
    text = f.read()

# トークン化（単語のリスト）
tokens = tagger.parse(text).strip().split()

# 単語の出現頻度をカウント
counter = Counter(tokens)
frequencies = sorted(counter.values(), reverse=True)

# ランクと頻度を log-log スケールでプロット
ranks = np.arange(1, len(frequencies) + 1)
freqs = np.array(frequencies)

plt.figure(figsize=(8, 6))
plt.loglog(ranks, freqs, marker='o', linestyle='none')
plt.xlabel("Rank (log scale)")
plt.ylabel("Frequency (log scale)")
plt.title("Zipf's Law: Word Frequency vs. Rank")
plt.grid(True, which="both", linestyle="--", linewidth=0.5)
plt.tight_layout()
plt.show()
```
